## JUNE RAINFALL BY FLOOD STATUS

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=dh, x='FLOODS', y='JUN', palette='Set2')
plt.title('Boxplot of June Rainfall by Flood Status')
plt.xlabel('Flood Status')
plt.ylabel('Annual Rainfall (mm)')
plt.show()
```

## ANALYSING ANNUAL RAINFALL BY EVERY YEAR

```python
plt.figure(figsize=(12, 6))
plt.bar(data["YEAR"],data[" ANNUAL RAINFALL"], color='pink')
plt.title('Annual Rainfall by Year')
plt.xlabel('Year')
plt.ylabel('Annual Rainfall (mm)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

 ### The year 2018 has the highiest rainfall 

## ANALYSING RAINFALL ON EACH MONTH IN 2018

```python
a=data[data["YEAR"]==2018].squeeze()
months=["JAN","FEB","MAR","APR","MAY","JUN","JUL","AUG","SEP","OCT",'NOV',"DEC"]
plt.figure(figsize=(10,5))
plt.bar(months,a[months],color="green")
plt.title("Rainfall on each month in 2018")
plt.xlabel("months")
plt.ylabel("rainfall recieved")
plt.show()
```

### Augest has the highest rainfall in the year 2018

## FLOOD OCCURANCE

```python
plt.figure(figsize=(15,7))
data["FLOODS"].value_counts().plot(kind="pie",colors=["skyblue","brown"],autopct="%1.1f%%",startangle=90)
plt.title("Flood occured")
plt.show()
```

### The rate of flood occured since 1901 to2018 is about 50.8%

## ANALYSING AVERAGE MONTHLY RANIFALL

```python
monthly_avrg=data[["JAN",'FEB','MAR',"APR",'MAY','JUN','JUL','AUG','SEP','OCT','NOV','DEC']].mean()
plt.figure(figsize=(14,6))
sns.barplot(x=monthly_avrg.index,y=monthly_avrg.values,color="brown")
plt.title("Average monthly rainfall in kerala-1901 to 2018")
plt.xlabel("Month")
plt.ylabel("Average Rainfall")
plt.show()
```

### July has the highest monthly rainfall from the year 1901 to 2018

## COMPARING MONSOON AND NON-MONSOON RAINFALL

```python
data['Monsoon_rainfall']=data[['JUN','JUL','AUG','SEP']].sum(axis=1)
data["Non_Monsoon_Rainfall"]=data[['JAN','FEB','MAR','APR','MAY','OCT','NOV','DEC']].sum(axis=1)
```

```python
plt.figure(figsize=(30,10))
data[['YEAR','Monsoon_rainfall','Non_Monsoon_Rainfall']].set_index("YEAR").plot(kind="line")
plt.title('Monsoon VS non-monsoon rainfall in Kerala')
plt.xlabel("Year")
plt.ylabel("Rainfall(mm)")
plt.grid(True)
plt.show()
```
